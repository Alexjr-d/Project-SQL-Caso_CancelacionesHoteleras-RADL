17. Hipótesis 1: ¿Los hijos aseguran el viaje? Escribe una consulta usando CASE WHEN que clasifique a la reserva como "Con Hijos" (si children > 0 o babies > 0) o "Sin Hijos". Agrupa por esta nueva categoría y calcula la tasa de cancelación (puedes sacar el promedio - AVG - de la columna is_canceled). ¿Qué segmento cancela más?

```python
%sql
SELECT
  CASE
    WHEN children > 0 OR babies > 0 THEN 'Con Hijos'
    ELSE 'Sin Hijos'
  END AS segmento,
  COUNT(*)                     AS total_reservas,
  SUM(is_canceled)             AS canceladas,
  ROUND(AVG(is_canceled), 2)   AS tasa_cancelacion
FROM cursodataacademy.project_hotel.tb_reservas
GROUP BY segmento
ORDER BY tasa_cancelacion DESC;
```

`Conclusion 17:` Los clientes sin hijos cancelan más, representado un 37% que los que viajan con hijos que en puntos porcentuales es 35% aunque la diferencia es de apenas 2 puntos porcentuales.
El segmento “Sin Hijos” concentra 110058 de las 119390 reservas, mientras que “Con Hijos” solo tiene 9332, así que la mayor parte del riesgo está en viajeros sin niños.
Por la poca diferencia, los hijos no aseguran el viaje de forma contundente. En resumen, la hipótesis se cumple, pero con un efecto débil.

18. Hipótesis 2: El impacto de las solicitudes especiales. Agrupa por la columna total_of_special_requests y calcula la tasa de cancelación (AVG(is_canceled)) para cada cantidad de solicitudes (0, 1, 2, etc.). Ordena por la cantidad de solicitudes de forma ascendente. ¿A medida que el cliente pide más cosas, cancela más o cancela menos?

```python
%sql
SELECT 
    total_of_special_requests,
    COUNT(*)                   AS total_reservas,
    SUM(is_canceled)           AS canceladas,
    ROUND(AVG(is_canceled), 2) AS tasa_cancelacion
FROM cursodataacademy.project_hotel.tb_reservas
GROUP BY total_of_special_requests
```

`Conclusion 18:` A medida que el cliente hace más solicitudes especiales, cancela menos y la tasa baja de 48% a 22%, y llega a 5% con 5 solicitudes.
El grupo sin solicitudes concentra 70318 reservas y 33556 cancelaciones, es decir, más de la mitad de todas las cancelaciones del hotel.
Pedir algo adicional es una señal clara de compromiso con el viaje, y el efecto es mucho más fuerte que el de viajar con hijos.

19. Hipótesis 3: ¿Cambiar la reserva es señal de compromiso? Al igual que en los ejercicios anteriores, utiliza CASE WHEN para segmentar la columna booking_changes en tres grupos: "Sin Cambios" (0), "1 a 3 cambios" y "Más de 3 cambios". Calcula la tasa de cancelación para cada grupo. ¿Quiénes tienen menor riesgo de cancelar?

```python
%sql
SELECT
  CASE
    WHEN booking_changes = 0 THEN 'Sin Cambios'
    WHEN booking_changes BETWEEN 1 AND 3 THEN '1 a 3 cambios'
    ELSE 'Más de 3 cambios'
  END                          AS segmento,
  COUNT(*)                     AS total_reservas,
  SUM(is_canceled)             AS canceladas,
  ROUND(AVG(is_canceled), 2)   AS tasa_cancelacion
FROM cursodataacademy.project_hotel.tb_reservas
GROUP BY segmento
ORDER BY tasa_cancelacion ASC;
```

`Conclusion 19:` Los clientes que modifican su reserva al menos una vez cancelan mucho menos representando un 16% con 1 a 3 cambios y 18% con más de 3, frente a 41% de quienes no hacen ninguno.
El segmento “Sin Cambios” reúne 101314 de las 119390 reservas y 41391 de las 44224 cancelaciones, es decir, cerca del 94% de todas las cancelaciones.
Por lo que, el grupo de menor riesgo es “1 a 3 cambios”

20. Hipótesis 4: El factor precio (adr). Agrupa las reservas por tipo de hotel (hotel). Para cada tipo, muestra la cantidad total de cancelaciones y el precio promedio (adr) solo de las reservas canceladas. En base a tu consulta, ¿en qué tipo de hotel el precio promedio de cancelación es más alto?

```python
%sql
SELECT 
    hotel,
    COUNT(*)           AS Total_Cancel,
    ROUND(AVG(adr), 2) AS AVG_Cancel
FROM cursodataacademy.project_hotel.tb_reservas
WHERE is_canceled LIKE '1'
GROUP BY hotel
```

La diferencia es de apenas S/1.10 por noche, así que el precio no distingue de forma clara a un hotel del otro en las cancelaciones.
